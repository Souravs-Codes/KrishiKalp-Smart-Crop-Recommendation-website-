# Soil Data Cleaning

### Objective

In this notebook, we will:

- Load the raw Soil Health dataset.
- Check for missing values and duplicates.
- Optimize memory usage.
- Convert appropriate columns to categorical data types.
- Save a cleaned version of the dataset for further processing.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/soil-nutrient-analysis.csv")
```

```python
print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")

df.info()
```

```python
df.isnull().sum()
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
memory_before = df.memory_usage(deep=True).sum() / (1024**2)

print(f"Memory Before Optimization: {memory_before:.2f} MB")
```

```python
##Reducing memory usage by converting categorical columns to category dtype

categorical_columns = [
    "year",
    "state_name",
    "district_name",
    "block_name",
    "village_name",
    "nutrient_type",
    "nutrient_name",
    "nutrient_level"
]

for column in categorical_columns:
    df[column] = df[column].astype("category")
```

```python
memory_after = df.memory_usage(deep=True).sum() / (1024**2)

print(f"Memory After Optimization : {memory_after:.2f} MB")

reduction = ((memory_before - memory_after) / memory_before) * 100

print(f"Memory Reduced By : {reduction:.2f}%")
```

### Save the Cleaned Dataset

```python
df.to_csv("../data/processed/soil_data_cleaned.csv", index=False)
```
